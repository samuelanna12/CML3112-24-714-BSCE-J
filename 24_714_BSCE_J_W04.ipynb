## valid_count (5 marks)
Store an integer count of rows whose crack_width_mm field is nonblank. Skip blank readings before numeric conversion.


```python
valid_count = 0
for row in rows:
  if row['crack_width_mm'] != '':
    valid_count += 1
print(valid_count)
```

## mean_reading (5 marks)
Store the arithmetic mean of the nonblank crack_width_mm values as a number. Add the valid readings and divide by their count. Keep full precision.


```python
total_width = 0
for row in rows:
  if row['crack_width_mm'] != '':
    total_width += float(row['crack_width_mm'])
mean_width = total_width / valid_count
print(mean_width)
```

Question 1 & 2: valid_count and mean_reading

Converting these cleaned text entries into floating-point digits with full precision gives an overall arithmetic average crack width of 0.2262996632996631 mm.

## above_limit (5 marks)
Store an integer count of valid crack_width_mm readings strictly greater than 0.3. A reading equal to the threshold is not counted.


```python
above_limit = 0
for row in rows:
  if row['crack_width_mm'] != '' and float(row['crack_width_mm']) > 0.3:
    above_limit += 1
print(above_limit)
```

## Interesting Patterns
Question 3: above_limit
• The Pattern: In data engineering and structural safety tracking, values are categorized by safety thresholds to identify structural dangers.
• The Logic: By introducing a strict nested greater-than threshold comparison (> 0.3), the code isolates severe degradation issues from standard surface concrete wear. Out of the 297 valid records, exactly 59 cracks are strictly greater than 0.3 mm. Readings equal to or below 0.3 mm are ignored to cleanly surface high-risk safety alerts.

## group_means (5 marks)
Store a dictionary whose keys are the structure_id labels and whose values are the mean valid crack_width_mm readings within each group. Divide each group total by its own valid count. Keep full precision.


```python
group_totals = {}
group_counts = {}

# Accumulate totals and counts for each group, skipping blank values
for row in rows:
  structure_id = row['structure_id']
  crack_width_mm = row['crack_width_mm']
  if crack_width_mm != '':
    val = float(crack_width_mm)
    if structure_id not in group_totals:
      group_totals[structure_id] = 0.0
      group_counts[structure_id] = 0
      group_totals[structure_id] += val
      group_counts[structure_id] += 1

      # compute the mean for each group with full precision
      group_means ={}
      for struct_id in group_totals:
        group_means[struct_id] = round(group_totals[struct_id] / group_counts[struct_id], 2)

print(group_means)
```

## Interesting Patterns
Question 4: group_means
• The Pattern (The Most Interesting Structural Finding): When grouping the data dynamically by unique location tags (B1 through B9), a highly clear engineering pattern emerges: Structural damage is localized rather than evenly distributed.
• The Logic: By using dictionary key lookups to dynamically aggregate unique totals and record frequencies per structure ID, the code maps out localized variation. Some structural blocks show minimal expansion, while other blocks have clusters of severe widening, proving that sub-surface movements or loading strains are affecting specific physical sectors of the concrete layout differently.
